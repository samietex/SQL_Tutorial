# SQL Tutorial

### A Hands-On Companion to the *All About Data & More* SQL Curriculum

Welcome. This is where the SQL fundamentals from the newsletter turn into something you actually run yourself. Work through the notebooks in order, execute every query, and don't just read the answer, get the wrong answer first if that's what it takes to understand why the right one works.

> **Who this is for:** anyone starting the Data Analytics, Analytics Engineering, or Data Engineering track from the [Start Here roadmap](https://aaboutdata.substack.com/p/start-here-from-beginner-to-job-ready), and anyone who just wants a real, hands-on SQL Server tutorial. You don't need to be subscribed to the newsletter to use this repo, but the two are built to work together.

## Getting Started

**Already comfortable with git, VS Code, and SQL Server?** Clone the repo, open it in VS Code with the mssql extension installed, restore AdventureWorks2022, and jump straight to `01 - SQL Basics`. Everything below this line is for anyone who isn't, and it's worth reading even if you're experienced, since it covers exactly what this repo needs, not a generic "how to use git" tutorial.

### What you're actually looking at

Every file ending in `.ipynb` is a Jupyter notebook: a document that mixes explanation, like the text you're reading right now, with actual SQL code you can run one step at a time and see the real result underneath it. That's the whole point of this repo. You don't just read about `SELECT`, you run it yourself against a real database.

### Reading vs. running

If you're viewing this on github.com, you're seeing a **preview**. GitHub can display a notebook's text and whatever results were already saved in it, but it cannot execute anything, there's no database behind the page you're looking at. To actually run a query yourself, you need to open the notebook somewhere connected to a real SQL Server, on your own computer.

### 1. Get the code onto your computer

If you're comfortable with git:
```
git clone https://github.com/samietex/SQL_Tutorial.git
```
If you've never used git before, that's fine, click the green **Code** button on this repo's GitHub page, then **Download ZIP**, and unzip it somewhere on your computer.

### 2. Install SQL Server and load the practice data

- Install [SQL Server Express Edition](https://www.microsoft.com/en-us/sql-server/sql-server-downloads), it's free.
- Install [SQL Server Management Studio (SSMS)](https://learn.microsoft.com/en-us/sql/ssms/download-sql-server-management-studio-ssms), the tool you'll use to load the sample database below.
- Download and restore the **AdventureWorks2022** sample database, most of these notebooks query it, and it doesn't come with SQL Server automatically. Microsoft's official instructions are here: [Install and configure AdventureWorks](https://learn.microsoft.com/en-us/sql/samples/adventureworks-install-configure). You want the OLTP `.bak` file, restored under the name `AdventureWorks2022`.

### 3. Install a tool that can actually run the notebooks

These notebooks use a SQL kernel, they need an app that knows how to connect a `.ipynb` file to a real SQL Server and run its cells against it. The easiest path:

- Install [Visual Studio Code](https://code.visualstudio.com/), it's free.
- Inside VS Code, install the **SQL Server (mssql)** extension by Microsoft (search "mssql" in the Extensions panel).
- Open this repo's folder in VS Code (`File > Open Folder`).
- Open `01 - SQL Basics (SELECT, WHERE, ORDER BY).ipynb`.
- The first time you run a cell, VS Code asks you to pick a kernel, choose the SQL kernel, then connect it to your local SQL Server instance (the mssql extension walks you through adding a connection if you don't have one saved yet).
- Run a cell with the ▷ button next to it, or `Shift+Enter`. A cell that queries data shows its result as a table right underneath it, that's a real result from your own database, not a canned example someone else already ran.

If you already use Azure Data Studio, that works too, it has the same SQL Notebooks feature built in.

## A Note on SQL Server vs. SQL

Everything here runs on SQL Server, because you need one real engine to actually install and run queries against, not three. But the language you're learning is standard SQL: `SELECT`, `WHERE`, `JOIN`, `GROUP BY`, CTEs, and window functions all work the same way on MySQL, PostgreSQL, and everywhere else, they're part of the ANSI SQL standard, not something Microsoft invented for SQL Server specifically.

The handful of places where SQL Server does its own thing, mostly around changing data and table structure in `08a`/`08b`, get called out inline as they come up, so you know what's universal and what's specific to this one engine before you ever touch a different one.
