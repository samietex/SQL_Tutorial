# SQL Tutorial

### A hands-on companion to the *All About Data & More* SQL curriculum

This is where the SQL fundamentals from the newsletter turn into something you run yourself. Each notebook follows one lesson: it builds the lesson's small dataset so every result matches the article exactly, walks through the ideas with queries you run and change, shows the mistakes that return a wrong answer without an error, gives you exercises with a separate solutions notebook, and ends by running the same ideas on a real company database with tens of thousands of rows.

> **Who this is for:** anyone starting the Data Analytics, Analytics Engineering, or Data Engineering track from the [Start Here roadmap](https://aaboutdata.substack.com/p/start-here-from-beginner-to-job-ready), and anyone who wants a careful, hands-on introduction to SQL. You don't need to be subscribed to the newsletter, and you don't need to have written any code before: `01 - What Is a Database` starts from zero.

## How these notebooks work

Every file ending in `.ipynb` is a **Jupyter notebook**: a document that mixes explanation, like the text you're reading, with code cells you run one at a time, each showing its result directly underneath. The first cell of every notebook loads a small helper, and after that, two kinds of cell run SQL:

- **`%%sql` cells** run against a small practice database that lives inside the notebook, built from the lesson's own tables. It uses **SQLite**, which comes with Python, so there's nothing to install, and it starts empty every time you open the notebook, so you can't break anything that matters.
- **`%%tsql` cells** run against **SQL Server** and **AdventureWorks2022**, Microsoft's sample database for an invented bike company, with 19,820 customers and 31,465 orders. Each notebook's last section uses it to show the lesson's ideas, and its traps, at real scale, along with the places where SQL Server behaves differently from SQLite.

Cells meant to show an error are marked `--expect-error`, so they print the error and let the notebook carry on.

## Three ways to use them

### 1. Read them on GitHub

Every notebook is saved with its results, so you can read the whole course on github.com without installing anything. GitHub shows a **preview**: it can't run a cell, so you can't change a query and see what happens, which is where most of the learning is.

### 2. Run them in Google Colab (recommended to start)

[Google Colab](https://colab.research.google.com/) runs notebooks in your browser for free with a Google account, with nothing to install. Every notebook has an **Open in Colab** button at the top.

1. Click **Open in Colab** on any notebook.
2. Run the first code cell (click it, then press **Shift + Enter**). It prints `Ready.` when the helper has loaded.
3. Work down the page, running each cell in order.
4. When you reach the AdventureWorks section, run its **Set up AdventureWorks** cell. It installs SQL Server and AdventureWorks inside your Colab session and prints each step; the first run takes about three to five minutes.

Colab gives you a fresh machine each session, so if you come back later, run the first cell again, and the AdventureWorks setup cell again before any `%%tsql` cell. Your own edits are kept only if you save a copy to your Google Drive (**File > Save a copy in Drive**).

### 3. Run them on your own computer

**Get the files.** With git, `git clone https://github.com/samietex/SQL_Tutorial.git`. Without it, click the green **Code** button on the repository page, then **Download ZIP**, and unzip it.

**Install Python and Jupyter.** Install [Python](https://www.python.org/downloads/) 3.10 or later, then, in a terminal inside the repository folder:

```
pip install -r requirements.txt
jupyter lab
```

JupyterLab opens in your browser. [VS Code with its Jupyter extension](https://code.visualstudio.com/docs/datascience/jupyter-notebooks) works too; pick a Python kernel when it asks. That's all the `%%sql` cells need.

**For the AdventureWorks sections**, you also need SQL Server with AdventureWorks restored:

- **On Windows:** install [SQL Server Developer edition](https://www.microsoft.com/en-us/sql-server/sql-server-downloads) (free for learning), the [ODBC Driver for SQL Server](https://learn.microsoft.com/en-us/sql/connect/odbc/download-odbc-driver-for-sql-server), and [SQL Server Management Studio](https://learn.microsoft.com/en-us/ssms/install/install), then restore the `AdventureWorks2022.bak` file by following Microsoft's [Install and configure AdventureWorks](https://learn.microsoft.com/en-us/sql/samples/adventureworks-install-configure). The notebooks then sign in with your Windows login automatically.
- **On macOS or Linux:** run SQL Server in Docker by following Microsoft's [SQL Server container quickstart](https://learn.microsoft.com/en-us/sql/linux/quickstart-install-connect-docker), restore the same backup into it, and tell the notebooks how to sign in by running this in a cell before the AdventureWorks setup cell, with the password you chose:

```
%env SQL_TUTORIAL_USER=sa
%env SQL_TUTORIAL_PASSWORD=<your password>
```

Then run each notebook's **Set up AdventureWorks** cell as normal: on your own machine it checks the connection and tells you what's missing if it can't reach SQL Server. `SQL_TUTORIAL_SERVER` (default `localhost`) and `SQL_TUTORIAL_DATABASE` (default `AdventureWorks2022`) can be set the same way if yours differ.

> **Coming from an earlier version of this repo?** The notebooks used to run on a SQL kernel in Azure Data Studio, which Microsoft [retired in February 2026](https://learn.microsoft.com/en-us/azure-data-studio/whats-happening-azure-data-studio). They now run on a standard Python kernel, which is what makes Colab and plain Jupyter possible.

## SQLite, SQL Server, and "SQL"

SQL is shared across many database engines, and each has its own **dialect**. The lessons run on SQLite because it needs no setup and lets every reader get identical results; the AdventureWorks sections run on SQL Server, because that's what many workplaces use and because seeing two engines side by side is the fastest way to learn which parts of SQL are universal. `SELECT`, `WHERE`, `JOIN`, `GROUP BY`, CTEs, and window functions work the same way almost everywhere. The differences, like `TOP 5` against `LIMIT 5`, how text compares, what `AVG` returns for whole numbers, or whether foreign keys are switched on, are called out where they come up, usually because they change an answer without an error.
