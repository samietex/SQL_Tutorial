## Data Query Language

### 1. The Mighty SELECT command

**Retrieving Data**

Use SELECT to fetch specific columns or all of them from a table. We will be considering the AdventureWorks database throughout this tutorial

**Example:**

````sql
USE AdventureWorks2022
````

````sql
SELECT CustomerID, TerritoryID, AccountNumber 
FROM Sales.Customer; -- Gets all columns from "Customer" table
````

**Column Alias**

Give temporary names to columns for readability.

**Example:**

````sql
SELECT DepartmentID AS "ID", Name AS "Department", ModifiedDate AS "Date"
FROM HumanResources.Department;
````

### 2. Filtering with WHERE: Finding the Right Fit

**WHERE Clause**

Refine your results with conditions.

**Example:**

````sql
SELECT NationalIDNumber, JobTitle
FROM HumanResources.Employee 
WHERE JobTitle = 'Research and Development Manager'; -- Only employees as research and development manager
````

Let's look at how we can use the `IN` operator to match against a list of values instead of chaining several `OR` conditions:

````sql
SELECT * FROM HumanResources.Employee
WHERE JobTitle IN ('Engineering Manager', 'Senior Tool');
````

````sql
SELECT COUNT(DISTINCT Title) AS "Count of Distinct Title"
FROM Person.Person;
````

**Operators**

Use <, >, =, LIKE, IN, etc., to compare and match data.

### 3. Ordering Made Easy: The ORDER BY Clause

**Organize Your Data**

Sort results in ascending or descending order based on one or more columns.

**Example:**
