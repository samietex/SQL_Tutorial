# 02 - Aggregation (GROUP BY, HAVING, COUNT/SUM/AVG)

Picks up right where `01 - SQL Basics` left off. Same AdventureWorks2022 database, same habit of asking what question a query is actually answering.

> **By the end of this notebook you'll be able to:** collapse many rows into one summary row per group with `GROUP BY`, and use `COUNT`, `SUM`, and `AVG` to summarize each group.

### 4. Grouping and Aggregating: Unveiling Patterns

**Group Functions**

Summarize data like counting, finding averages, or finding maximum/minimum values for groups.

**Example:**

`GROUP BY` collapses every row that shares the same value in a column into a single summary row. Here, every person in `Person.Person` gets bucketed by their `PersonType`, and `COUNT(*)` counts how many rows landed in each bucket:

<p align="center">
  <img src="graphics/02_group_by_collapse.png" width="400" alt="What GROUP BY actually does to your rows">
</p>

*Diagram source: `graphics/02_group_by_collapse.mmd`*

```sql
SELECT PersonType, COUNT(*) AS "Person Count" 
FROM Person.Person
GROUP BY PersonType;
```
