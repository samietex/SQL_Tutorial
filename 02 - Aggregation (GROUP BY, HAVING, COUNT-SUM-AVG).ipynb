### 4. Grouping and Aggregating: Unveiling Patterns

**Group Functions**

Summarize data like counting, finding averages, or finding maximum/minimum values for groups.

**Example:**

`GROUP BY` collapses every row that shares the same value in a column into a single summary row. Here, every person in `Person.Person` gets bucketed by their `PersonType`, and `COUNT(*)` counts how many rows landed in each bucket:

<p align="center">
  <img src="graphics/02_group_by_collapse.png" width="400" alt="What GROUP BY actually does to your rows">
</p>

```sql
SELECT PersonType, COUNT(*) AS "Person Count" 
FROM Person.Person
GROUP BY PersonType;
```

**SUM and AVG**

`COUNT` tells you how many rows are in a group. `SUM` and `AVG` summarize a numeric column across that same group, total it up or average it out. Same `GROUP BY`, different function.

**Example:**

Every row in `Sales.SalesOrderHeader` is one order. Grouped by `CustomerID`, `SUM(TotalDue)` adds up everything a customer has ever spent, and `AVG(TotalDue)` shows what a typical order from them looks like:

```sql
SELECT
    CustomerID,
    SUM(TotalDue) AS "Total Spent",
    AVG(TotalDue) AS "Average Order Value",
    COUNT(*) AS "Number of Orders"
FROM Sales.SalesOrderHeader
GROUP BY CustomerID
ORDER BY "Total Spent" DESC;
```

### 5. HAVING: filtering the groups themselves

**Group Filters**

`WHERE` filters individual rows, and it runs *before* `GROUP BY` collapses them. By the time `SUM(TotalDue)` exists as a value, `WHERE` has already finished, there's no row left for it to filter, only a group. `HAVING` is the clause that runs after grouping, so it's the one that can filter on `SUM(TotalDue)`, `COUNT(*)`, or any other aggregate. Same idea as the query execution order from `01 - SQL Basics`, `HAVING` just slots in after `GROUP BY` instead of after `FROM`.

**Example:**

Same query as above, but now only keep customers whose total spend clears $50,000:

```sql
SELECT
    CustomerID,
    SUM(TotalDue) AS "Total Spent"
FROM Sales.SalesOrderHeader
GROUP BY CustomerID
HAVING SUM(TotalDue) > 50000
ORDER BY "Total Spent" DESC;
-- HAVING runs after the grouping and SUM happen, WHERE SUM(TotalDue) > 50000 would fail,
-- there's no SUM(TotalDue) column to filter on until the rows are already grouped
```

Hold onto that last query. `04 - Subqueries to CTEs` wraps this *exact* pattern in a `WITH high_value_customers AS (...)` block and builds on it, once `GROUP BY` / `HAVING` feels natural, that notebook is mostly about giving a query like this one a name.
